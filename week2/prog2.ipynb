```python
#3.example code on match case
day=int(input("enter the day:"))
match day:
  case1:
    print("sunday")
  case2:
    print("monday")
  case3:
    print("tuesday")
  case4:
    print("wednesday")
  case5:
    print("thursday")
  case6:
    print("friday")
  case7:
    print("saturday")
  case_:
    print("invalid")
```

```python
#reverse of a given number
#318
n=int(input("enter the number:"))
rev=0
while(n!=0):
  rem=n%10
  rev=rev*10+rem
  n=n//10
print(rev)
```

```python
# program to find sum of digits of a given number
n=int(input("enter the number:"))
sum=0
while(n!=0):
  rem=n%10
  sum=sum+rem
  n=n//10
print(sum)
```
