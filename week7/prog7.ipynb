```python
l=[int(i) for i in input().split()]
s=set(l)
print(s)
print(len(l)-len(s))
```

```python
l=[int(i) for i in input().split()]
l2=[]
for i in l:
  if l.count(i)>1:
    l2.append(i)
print(l2)
s=set(l2)
print(s)
print(len(s))
```

```python
n=int(input())
l=[int(i) for i in input().split()]
l2=set(l)
print(l2)
element=set(range(1,n+1))
print(element)
print(element-l2)
```

```python
key=[i for i in input().split()]
value=[j for j in input().split()]
d={i:j for i,j in zip(key,value)}
k={m for m in input().split()}
if k.issubset(d):
  print("yes")
else:
  print("no")
```
